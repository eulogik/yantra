# Yantra — Recovery: SFT on 66 fixed args + replay → export → guarded eval

Upload to Colab (T4 GPU) → **Runtime → Run all**. ~30 min total.

**Why SFT, not DPO:** the diagnostic proved DPO-on-66-pairs collapsed arg generation — the model now emits one blob param keyed by the *tool name* (`name="check_email_spam">{...json...}`) instead of per-key `<param>`s. Contrastive pressure on 66 pairs destroyed the keys; SFT only shows correct completions, so it cannot collapse this way. To prevent forgetting, the 66 fixes are mixed with 200 replay rows from the original `dtsa_sft.jsonl` (same Stage-3 format), 1 epoch, LR 1e-5 (10× below Stage 3).

**Safety:** Cell 5 promotes to `yantra_results.json` **only if PAS beats 0.6746**. Your official best is never overwritten by a worse run again. The bad DPO2 artifacts are left untouched (not deleted, just unused).

```
# @title 2 — Shared utils (Router V2 + parsers + metrics, same as 0.6746 run)
import json, re, math
from collections import Counter

ACTION_END = '<action_end/>'
DTSA_BIND = re.compile(r'<bind\s+tool="([^"]+)"\s*/>', re.S)
DTSA_ARGS = re.compile(r'<args>(.*?)</args>', re.S)
DTSA_PARAM = re.compile(r'<param\s+name="([^"]+)"\s*>(.*?)</param>', re.S)
def _unesc(v): return v.replace('&amp;','&').replace('&lt;','<').replace('&gt;','>')
def to_dtsa(name, arguments):
    lines = ['<bind tool="%s"/>' % name, '<args>']
    for k,v in arguments.items():
        v = '' if v is None else str(v)
        v = v.replace('&','&amp;').replace('<','&lt;').replace('>','&gt;')
        lines.append('  <param name="%s">%s</param>' % (k,v))
    lines += ['</args>', ACTION_END]
    return '\n'.join(lines)
def bind_prefix(name): return f'<bind tool="{name}"/>\n'
def dtsa_args_block(name, args): return '\n'.join(to_dtsa(name,args).splitlines()[1:])

def parse_dtsa(text):
    binds = list(DTSA_BIND.finditer(text))
    if not binds: return None
    ends = list(re.finditer(re.escape(ACTION_END), text))
    stopped = bool(ends) and text[ends[-1].end():].strip() == ''
    for j in range(len(binds)-1, -1, -1):
        lb = binds[j]
        seg_end = binds[j+1].start() if j+1 < len(binds) else len(text)
        am = DTSA_ARGS.search(text[lb.end():seg_end])
        if not am: continue
        body = am.group(1); args = {}
        for pm in DTSA_PARAM.finditer(body):
            args[pm.group(1)] = _unesc(pm.group(2).strip())
        if not args:
            for lm in re.finditer(r'name="([^"]+)"\s*>[ \t]*(.*)', body):
                args[lm.group(1)] = _unesc(lm.group(2).strip())
        return {'tool': lb.group(1), 'args': args, 'stopped_clean': stopped}
    return {'tool': binds[0].group(1), 'args': {}, 'stopped_clean': stopped}

def _norm(a):
    if isinstance(a, str):
        s = a.strip()
        if s.lower() in ('true','false'): return s.lower()=='true'
        try: return int(s) if '.' not in s else float(s)
        except ValueError: return s
    return a
def avail_names(tools): return {t.get('function', t).get('name') for t in tools}
def evaluate_case(text, case, mode='dtsa'):
    tools = case['tools']; gold = case['gold']; gold0 = gold[0]
    parsed = parse_dtsa(text)
    if not parsed: return {'parseable':0,'valid_name':0,'expected_name':0,'exact_args':0,'arg_key_overlap':0,'stopped_cleanly':0}
    valid = parsed['tool'] in avail_names(tools)
    expected = parsed['tool'] == gold0['name']
    gk = set(gold0['arguments'].keys()); pk = set(parsed['args'].keys())
    overlap = len(gk & pk)/len(gk) if gk else 1.0
    exact = (set(parsed['args'].keys()) == gk) and all(_norm(gold0['arguments'][k]) == _norm(parsed['args'][k]) for k in gk) if gk else bool(parsed)
    return {'parseable':1,'valid_name':int(valid),'expected_name':int(expected),'exact_args':int(exact),'arg_key_overlap':round(overlap,4),'stopped_cleanly':int(parsed['stopped_clean'])}
def pas(summary, recovery=0.0, multiturn=0.0):
    comps = [summary.get(k,0.0) for k in ('parseable','valid_name','expected_name','exact_args','arg_key_overlap','stopped_cleanly')] + [recovery, multiturn]
    return round(sum(comps)/len(comps), 4)

STOP = ['\n<bind', '<tool_result>', '<user>', '</calls>', '<tool_error>', '<reflect>']
print('utils loaded. STOP =', STOP)
```

```
# @title 3 — Build SFT dataset: 66 fixes + 200 replay (Stage-3 format, ~1 min)
import random

# 3a. Fixes: prompt+chosen from the mined file (prompt already carries the gold bind).
fixes = [json.loads(l) for l in open(ART/'dpo_round2_from_eval.jsonl') if l.strip()]
fix_rows = [{'text': p['prompt'] + p['chosen'] + '\n'} for p in fixes]
print(f'fixes: {len(fix_rows)}')

# 3b. Replay: 200 random Stage-1 rows rebuilt in Stage-3 format (gold bind + args-only).
sft_rows = [json.loads(l) for l in open(ART/'dtsa_sft.jsonl') if l.strip()]
rng = random.Random(11)
replay_src = rng.sample(sft_rows, min(200, len(sft_rows)))
replay_rows = []
for r in replay_src:
    name = r['gold'][0]['name']
    prompt = (f"<user>{r['query']}</user>\n<tools>"
              f"{json.dumps([t.get('function',t) for t in r['tools']], ensure_ascii=False)}"
              f"</tools>\n<calls>" + bind_prefix(name))
    replay_rows.append({'text': prompt + dtsa_args_block(name, r['gold'][0]['arguments']) + '\n'})
print(f'replay: {len(replay_rows)} from {len(sft_rows)} stage-1 rows')

data = fix_rows + replay_rows
rng.shuffle(data)
(ART/'sft_round2_replay.jsonl').write_text(
    '\n'.join(json.dumps(d, ensure_ascii=False) for d in data))
print(f'total SFT rows: {len(data)} → {ART/"sft_round2_replay.jsonl"}')
print('sample fix len:', len(fix_rows[0]['text']), '| sample replay len:', len(replay_rows[0]['text']))
```

```
# @title 4 — SFT Round-2 (1 epoch, LR 1e-5, ~5-10 min on T4)
import warnings, logging
warnings.filterwarnings("ignore")
logging.getLogger("transformers").setLevel(logging.ERROR)
from unsloth import FastLanguageModel
from trl import SFTTrainer
from transformers import TrainingArguments
from datasets import Dataset

if BASE is None or not Path(BASE).exists():
    raise FileNotFoundError('No base adapters found. Train the main pipeline first.')
print('Base:', BASE)

rows = [json.loads(l) for l in open(ART/'sft_round2_replay.jsonl') if l.strip()]
print(f'SFT rows: {len(rows)}')
ds = Dataset.from_list(rows)
out = ART/'stage7_sft_round2'
out.mkdir(parents=True, exist_ok=True)

# Same load pattern as pipeline Stages 4/5 (adapters dir + identical LoRA config).
model, tokenizer = FastLanguageModel.from_pretrained(str(BASE), max_seq_length=2048)
model = FastLanguageModel.get_peft_model(
    model, r=64, lora_alpha=128, lora_dropout=0.05,
    target_modules=["q_proj","k_proj","v_proj","o_proj","gate_proj","up_proj","down_proj"])

trainer = SFTTrainer(
    model=model, tokenizer=tokenizer, train_dataset=ds,
    formatting_func=lambda x: x["text"],
    args=TrainingArguments(per_device_train_batch_size=4, gradient_accumulation_steps=4,
        warmup_ratio=0.03, num_train_epochs=1, learning_rate=1e-5, fp16=True,
        logging_steps=10, save_strategy="no", output_dir=str(out)))
trainer.train()
model.save_pretrained(str(out/"adapters"))
tokenizer.save_pretrained(str(out/"adapters"))
print('SFT Round-2 done →', out/"adapters")
```

```
# @title 5 — Merge + export Q4_K_M GGUF (~10-15 min), then free GPU
import shutil, gc
try:
    import torch
except Exception:
    pass

try:
    model
    print('Using in-memory trained model')
except NameError:
    from unsloth import FastLanguageModel
    _ad = ART/'stage7_sft_round2'/'adapters'
    if not _ad.exists():
        raise FileNotFoundError(f'{_ad} not found — run Cell 4 first.')
    print('Reloading', _ad)
    model, tokenizer = FastLanguageModel.from_pretrained(str(_ad), max_seq_length=2048)

for d in (Path('/content/stage8_sft2_q4'), Path('/content/stage8_sft2_q4_gguf')):
    if d.exists():
        shutil.rmtree(d)
print('Exporting Q4_K_M ...')
model.save_pretrained_gguf('/content/stage8_sft2_q4', tokenizer,
                            quantization_method='q4_k_m')

NEW_GGUF = None
for d in (Path('/content/stage8_sft2_q4_gguf'), Path('/content/stage8_sft2_q4')):
    if d.exists():
        g = sorted(d.glob('*Q4_K_M*.gguf')) or sorted(d.glob('*.gguf'))
        if g:
            NEW_GGUF = g[0]; break
assert NEW_GGUF is not None, 'GGUF export produced no file'
print(f'Exported: {NEW_GGUF} ({NEW_GGUF.stat().st_size/1024/1024:.0f} MB)')

try:
    dst = ART/'stage8_sft2_q4_gguf'
    if dst.exists():
        shutil.rmtree(dst)
    shutil.copytree(str(NEW_GGUF.parent), str(dst))
    print('Mirrored to Drive:', dst)
except Exception as e:
    print('Drive mirror skipped (non-fatal):', e)

for _n in ('trainer', 'model', 'tokenizer'):
    try:
        del globals()[_n]
    except KeyError:
        pass
gc.collect()
try:
    torch.cuda.empty_cache()
except Exception:
    pass
print('GPU freed. NEW_GGUF =', NEW_GGUF)
```

```
# @title 6 — Eval new GGUF + GUARDED promote (only if beats 0.6746, ~2 min)
from llama_cpp import Llama
import time

_gg = None
try:
    _gg = NEW_GGUF  # noqa: F821 — set in Cell 5
except NameError:
    _gg = None
if _gg is None or not Path(str(_gg)).exists():
    for d in (Path('/content/stage8_sft2_q4_gguf'), Path('/content/stage8_sft2_q4'),
              ART/'stage8_sft2_q4_gguf'):
        if d.exists():
            g = sorted(d.glob('*Q4_K_M*.gguf')) or sorted(d.glob('*.gguf'))
            if g:
                _gg = g[0]; break
if _gg is None:
    raise FileNotFoundError('No SFT2 GGUF found — run Cell 5 first.')
print('Evaluating:', _gg)
llm3 = Llama(model_path=str(_gg), n_ctx=4096,
             n_gpu_layers=(-1 if HAS_CUDA else 0), verbose=False)

cases = [json.loads(l) for l in open(ART/'toolace_300.jsonl') if l.strip()]
toks = lambda s: set(re.findall(r'[a-z0-9_]+', s.lower()))
_docs = [toks((t.get('function',t).get('name','')+' '+t.get('function',t).get('description',''))) for c in cases for t in c['tools']]
_df = Counter(w for d in _docs for w in d); _N = max(len(_docs),1)
_idf = lambda w: math.log(_N/(1+_df[w]))
def cgrams(s, n=3):
    s = re.sub(r'[^a-z0-9 ]','',s.lower())
    return {s[i:i+n] for i in range(max(len(s)-n+1,1))}
def route(q, tools):
    qt = toks(q); qg = cgrams(q)
    best, bn = -1e18, None
    for t in tools:
        f = t.get('function', t); nm = f.get('name') or ''
        d = toks(nm+' '+(f.get('description') or ''))
        s1 = sum(_idf(w) for w in qt & d)/(math.sqrt(sum(_idf(w) for w in d)+1e-6))
        ng = cgrams(nm)
        s2 = len(qg & ng)/(math.sqrt(len(qg)*len(ng))+1)
        if s1 + 2.0*s2 > best: best, bn = s1 + 2.0*s2, nm
    return bn

per = []
t0 = time.time()
for i, case in enumerate(cases):
    bind = f'<bind tool="{route(case["query"], case["tools"])}"/>\n'
    prompt = f"<user>{case['query']}</user>\n<tools>{json.dumps([t.get('function',t) for t in case['tools']], ensure_ascii=False)}</tools>\n<calls>{bind}"
    out = llm3(prompt, max_tokens=768, temperature=0.0, stop=STOP)
    per.append({'id': i, **evaluate_case(bind + (out['choices'][0]['text'] or ''), case)})
    if (i+1) % 50 == 0 or i == len(cases)-1:
        agg_t = {}
        for mm in per:
            for k, v in mm.items():
                if k == 'id': continue
                agg_t[k] = agg_t.get(k, 0.0) + v
        st = {k: round(v/len(per), 4) for k, v in agg_t.items()}
        print(f"  eval {i+1}/{len(cases)}  pas={pas(st):.4f}  exact={st.get('exact_args',0):.3f}  {time.time()-t0:.0f}s", flush=True)

agg = {}
for m in per:
    for k, v in m.items():
        if k == 'id': continue
        agg[k] = agg.get(k, 0.0) + v
summary = {k: round(v/len(cases), 4) for k, v in agg.items()}
score = pas(summary)
(ART/'yantra_results_sft2.json').write_text(
    json.dumps({'summary': summary, 'pas': score, 'per_case': per}, indent=2))
old = json.loads((ART/'yantra_results.json').read_text())
print('\n' + '='*50)
print(f"CURRENT OFFICIAL: PAS = {old.get('pas')}")
print(f"SFT2 CANDIDATE:   PAS = {score}  {summary}")
print(f"Δ exact_args: {old['summary']['exact_args']:.4f} → {summary['exact_args']:.4f} "
      f"({summary['exact_args']-old['summary']['exact_args']:+.4f})")
if score > float(old.get('pas', 0)):
    bak = ART/f"yantra_results_backup_{time.strftime('%Y%m%dT%H%M%S')}.json"
    shutil.copy2(str(ART/'yantra_results.json'), str(bak))
    shutil.copy2(str(ART/'yantra_results_sft2.json'), str(ART/'yantra_results.json'))
    print(f'IMPROVED → promoted (backup: {bak.name})')
else:
    print('NOT better → official 0.6746 KEPT. Nothing overwritten.')
```
